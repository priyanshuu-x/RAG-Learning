```python
## Text Loader 
from langchain_community.document_loaders import TextLoader
loader = TextLoader("data/text_files/ai_intro.txt",encoding="utf-8")
```

```python
documents=loader.load()
print(type(documents))
print(documents)
print(len(documents))
print(documents[0].page_content[:200])
```

Text Splitting


```python
from langchain_text_splitters import(
    CharacterTextSplitter,
    RecursiveCharacterTextSplitter,
    TokenTextSplitter
)

print(documents)
```

```python
## Method 1 - Character Text Splitter
text=documents[0].page_content
text
```

```python
print("---CHARACTER TEXT SPLITTER---")
char_splitter = CharacterTextSplitter(
    separator="\n",
    chunk_size=200,
    chunk_overlap=20,
    length_function=len
)

char_chunks=char_splitter.split_text(text)
print(f"Created {len(char_chunks)} chunks")
print(char_chunks[0])
print(char_chunks[1])
print(char_chunks[2])
```

```python
simple_text = "Python is a high-level, interpreted programming language known for its simple syntax, readability, and versatility. Created by Guido van Rossum and first released in 1991, Python has become one of the world's most popular programming languages. It supports multiple programming paradigms, including object-oriented, procedural, and functional programming. Python is widely used in web development, data science, artificial intelligence, machine learning, automation, scientific computing, and software development. Its extensive standard library and large ecosystem of third-party packages make it an excellent choice for beginners as well as experienced developers. With strong community support and cross-platform compatibility, Python continues to play a major role in modern technology and innovation."

splitter = RecursiveCharacterTextSplitter(
    separators=[" "],
    chunk_size=100,
    chunk_overlap=30,
    length_function=len
)

chunks = splitter.split_text(simple_text)

print(f"Created {len(chunks)} chunks")

for i in range(len(chunks)-1):
    print(f"chunk {i+1}: '{chunks[i]}'")
    print(f"chunk {i+2}: '{chunks[i+1]}'")

    print()
```
